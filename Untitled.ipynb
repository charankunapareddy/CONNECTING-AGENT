```python
from flask import Flask, request, redirect, url_for, flash, render_template_string, abort
import sqlite3
from pathlib import Path

# Create a folder for our website and its database
PROJECT = Path.home() / "Desktop" / "AgentConnect"
PROJECT.mkdir(parents=True, exist_ok=True)
DB_PATH = PROJECT / "agents.db"

app = Flask(__name__)
app.secret_key = "development-key-change-before-deployment"


# ---------- DATABASE ----------

def get_db():
    db = sqlite3.connect(DB_PATH)
    db.row_factory = sqlite3.Row
    return db


def initialize_database():
    with get_db() as db:
        db.execute("""
            CREATE TABLE IF NOT EXISTS agents (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                city TEXT NOT NULL,
                languages TEXT NOT NULL,
                experience INTEGER NOT NULL,
                phone TEXT NOT NULL,
                email TEXT NOT NULL,
                services TEXT NOT NULL,
                bio TEXT NOT NULL,
                approved INTEGER NOT NULL DEFAULT 0
            )
        """)

        db.execute("""
            CREATE TABLE IF NOT EXISTS inquiries (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                agent_id INTEGER NOT NULL,
                client_name TEXT NOT NULL,
                client_email TEXT NOT NULL,
                message TEXT NOT NULL,
                created_at TEXT DEFAULT CURRENT_TIMESTAMP
            )
        """)

        count = db.execute(
            "SELECT COUNT(*) FROM agents"
        ).fetchone()[0]

        if count == 0:
            samples = [
                (
                    "Ananya Rao", "Vijayawada",
                    "Telugu, English", 8,
                    "9000000001", "ananya@example.com",
                    "Family protection, Retirement planning",
                    "Helps families understand insurance options "
                    "and plan for their financial future.", 1
                ),
                (
                    "Rahul Sharma", "Hyderabad",
                    "English, Hindi, Telugu", 6,
                    "9000000002", "rahul@example.com",
                    "Savings plans, Child future planning",
                    "Explains insurance features to help clients "
                    "make informed decisions.", 1
                ),
                (
                    "Priya Kumar", "Visakhapatnam",
                    "Telugu, English", 4,
                    "9000000003", "priya@example.com",
                    "Life cover, Retirement planning",
                    "Provides guidance on insurance questions "
                    "and appointment-based consultations.", 1
                )
            ]

            db.executemany("""
                INSERT INTO agents
                (name, city, languages, experience, phone,
                 email, services, bio, approved)
                VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?)
            """, samples)


initialize_database()


# ---------- FRONTEND DESIGN ----------

STYLE = """
* { box-sizing: border-box; }
body {
    margin: 0;
    background: #f7f6f0;
    color: #19352f;
    font-family: Arial, sans-serif;
}
nav {
    padding: 22px 7%;
    background: white;
    display: flex;
    justify-content: space-between;
    align-items: center;
    border-bottom: 1px solid #e2e8df;
}
.brand { font-size: 23px; font-weight: bold; color: #28594b; }
nav a { color: #28594b; text-decoration: none; margin-left: 18px; }
.hero {
    padding: 65px 8%;
    background: linear-gradient(120deg, #f7f6f0, #e4eee4);
}
.eyebrow {
    color: #b18a45;
    font-size: 12px;
    letter-spacing: 2px;
    font-weight: bold;
}
h1 { font-size: clamp(35px, 5vw, 56px); margin: 15px 0; }
h2 { font-size: 30px; }
p { line-height: 1.7; color: #65766e; }
.search, .form-card {
    background: white;
    border: 1px solid #e0e7df;
    border-radius: 14px;
    padding: 22px;
    margin-top: 25px;
}
.search {
    display: flex;
    flex-wrap: wrap;
    gap: 12px;
}
input, textarea {
    padding: 12px;
    width: 100%;
    border: 1px solid #d7e0d7;
    border-radius: 7px;
    margin-top: 7px;
    font: inherit;
}
.search label { flex: 1; min-width: 180px; }
button, .button {
    background: #28594b;
    color: white;
    padding: 13px 18px;
    border: 0;
    border-radius: 7px;
    cursor: pointer;
    text-decoration: none;
    display: inline-block;
}
.section { padding: 45px 8%; }
.grid {
    display: grid;
    grid-template-columns: repeat(auto-fit, minmax(245px, 1fr));
    gap: 20px;
}
.card {
    padding: 24px;
    background: white;
    border: 1px solid #e0e7df;
    border-radius: 13px;
}
.avatar {
    width: 52px;
    height: 52px;
    display: grid;
    place-items: center;
    border-radius: 50%;
    background: #e0eee0;
    color: #28594b;
    font-size: 24px;
}
.tag {
    display: inline-block;
    background: #edf4ed;
    color: #28594b;
    padding: 6px 9px;
    margin: 4px 2px;
    border-radius: 20px;
    font-size: 12px;
}
.form-wrap { max-width: 700px; margin: 45px auto; padding: 0 20px; }
.form-card label { display: block; margin: 14px 0; }
.notice {
    margin: 20px 8%;
    padding: 14px;
    border-radius: 8px;
    background: #e1f2e4;
    color: #24563c;
}
footer {
    background: white;
    border-top: 1px solid #e0e7df;
    padding: 25px;
    text-align: center;
    font-size: 12px;
}
@media(max-width:600px) {
    nav { padding: 18px 5%; flex-wrap: wrap; gap: 10px; }
    nav a { margin-left: 8px; font-size: 13px; }
    .hero, .section { padding: 35px 5%; }
}
"""

BASE = """
<!doctype html>
<html lang="en">
<head>
<meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<title>{{ title }} | AgentConnect</title>
<style>""" + STYLE + """</style>
</head>
<body>
<nav>
  <a class="brand" href="{{ url_for('home') }}">AgentConnect</a>
  <div>
    <a href="{{ url_for('home') }}">Find agents</a>
    <a href="{{ url_for('register') }}">Join as agent</a>
  </div>
</nav>

{% for message in get_flashed_messages() %}
<div class="notice">{{ message }}</div>
{% endfor %}

{{ content|safe }}

<footer>
  <b>AgentConnect</b>
  <p>Independent educational directory. Not an official LIC website.</p>
  <p>Sample agent profiles are fictional and for demonstration only.</p>
</footer>
</body>
</html>
"""


def page(title, content, **context):
    return render_template_string(
        BASE, title=title, content=content, **context
    )


# ---------- HOME AND SEARCH ----------

@app.route("/")
def home():
    keyword = request.args.get("q", "").strip()
    city = request.args.get("city", "").strip()

    sql = "SELECT * FROM agents WHERE approved = 1"
    params = []

    if keyword:
        sql += " AND (name LIKE ? OR services LIKE ? OR languages LIKE ?)"
        value = "%" + keyword + "%"
        params.extend([value, value, value])

    if city:
        sql += " AND city LIKE ?"
        params.append("%" + city + "%")

    sql += " ORDER BY name"

    with get_db() as db:
        agents = db.execute(sql, params).fetchall()

    content = """
    <section class="hero">
      <div class="eyebrow">A CLEARER WAY TO CONNECT</div>
      <h1>Find guidance for<br>life's big decisions.</h1>
      <p>Discover insurance professionals by location, language,
         and the services you need.</p>

      <form class="search" method="get">
        <label>Service, name or language
          <input name="q" value="{{ keyword }}"
                 placeholder="Retirement planning, Telugu">
        </label>
        <label>City
          <input name="city" value="{{ city }}"
                 placeholder="Vijayawada">
        </label>
        <button type="submit">Search agents →</button>
      </form>
    </section>

    <section class="section">
      <div class="eyebrow">THE DIRECTORY</div>
      <h2>Meet your next advisor</h2>
      <p>{{ agents|length }} matching profile(s)</p>

      <div class="grid">
      {% for a in agents %}
        <article class="card">
          <div class="avatar">{{ a['name'][0] }}</div>
          <h3>{{ a['name'] }}</h3>
          <p>⌖ {{ a['city'] }}</p>
          <span class="tag">✓ Reviewed listing</span>
          <span class="tag">{{ a['experience'] }} years' experience</span>
          <p>{{ a['bio'] }}</p>
          <p><b>Languages:</b> {{ a['languages'] }}</p>
          <p><b>Services:</b> {{ a['services'] }}</p>
          <a class="button"
             href="{{ url_for('profile', agent_id=a['id']) }}">
             View profile →
          </a>
        </article>
      {% else %}
        <p>No agents found. Try another city or keyword.</p>
      {% endfor %}
      </div>
    </section>
    """
    return page("Find an agent", content,
                agents=agents, keyword=keyword, city=city)


# ---------- AGENT PROFILE AND CLIENT INQUIRIES ----------

@app.route("/agent/<int:agent_id>")
def profile(agent_id):
    with get_db() as db:
        a = db.execute(
            "SELECT * FROM agents WHERE id=? AND approved=1",
            (agent_id,)
        ).fetchone()

    if a is None:
        abort(404)

    content = """
    <section class="section">
      <a href="{{ url_for('home') }}">← Back to agents</a>
      <div class="form-card">
        <div class="avatar">{{ a['name'][0] }}</div>
        <h1>{{ a['name'] }}</h1>
        <p>⌖ {{ a['city'] }}</p>
        <span class="tag">Reviewed listing</span>
        <p>{{ a['bio'] }}</p>
        <p><b>Experience:</b> {{ a['experience'] }} years</p>
        <p><b>Languages:</b> {{ a['languages'] }}</p>
        <p><b>Services:</b> {{ a['services'] }}</p>

        <h2>Request a conversation</h2>
        <form method="post"
              action="{{ url_for('contact', agent_id=a['id']) }}">
          <label>Your name
            <input name="client_name" maxlength="100" required>
          </label>
          <label>Email address
            <input name="client_email" type="email"
                   maxlength="150" required>
          </label>
          <label>How can the agent help?
            <textarea name="message" maxlength="2000"
                      rows="4" required></textarea>
          </label>
          <button type="submit">Send inquiry →</button>
        </form>
      </div>
    </section>
    """
    return page(a["name"], content, a=a)


@app.route("/agent/<int:agent_id>/contact", methods=["POST"])
def contact(agent_id):
    name = request.form.get("client_name", "").strip()
    email = request.form.get("client_email", "").strip()
    message = request.form.get("message", "").strip()

    if not name or not email or not message:
        flash("Please fill in all fields.")
        return redirect(url_for("profile", agent_id=agent_id))

    if len(name) > 100 or len(email) > 150 or len(message) > 2000:
        flash("One or more fields are too long.")
        return redirect(url_for("profile", agent_id=agent_id))

    with get_db() as db:
        agent = db.execute(
            "SELECT id FROM agents WHERE id=? AND approved=1",
            (agent_id,)
        ).fetchone()

        if agent is None:
            abort(404)

        db.execute("""
            INSERT INTO inquiries
            (agent_id, client_name, client_email, message)
            VALUES (?, ?, ?, ?)
        """, (agent_id, name, email, message))

    flash("Inquiry saved in the local database.")
    return redirect(url_for("profile", agent_id=agent_id))


# ---------- AGENT REGISTRATION ----------

@app.route("/register", methods=["GET", "POST"])
def register():
    if request.method == "POST":
        name = request.form.get("name", "").strip()
        city = request.form.get("city", "").strip()
        languages = request.form.get("languages", "").strip()
        experience_text = request.form.get("experience", "0").strip()
        phone = request.form.get("phone", "").strip()
        email = request.form.get("email", "").strip()
        services = request.form.get("services", "").strip()
        bio = request.form.get("bio", "").strip()

        try:
            experience = int(experience_text)
            if not 0 <= experience <= 70:
                raise ValueError
        except ValueError:
            flash("Experience must be a whole number from 0 to 70.")
            return redirect(url_for("register"))

        if not all([name, city, languages, phone, email, services, bio]):
            flash("Please complete all fields.")
            return redirect(url_for("register"))

        with get_db() as db:
            db.execute("""
                INSERT INTO agents
                (name, city, languages, experience, phone,
                 email, services, bio, approved)
                VALUES (?, ?, ?, ?, ?, ?, ?, ?, 0)
            """, (name, city, languages, experience, phone,
                  email, services, bio))

        flash("Registration received. Your profile is pending review.")
        return redirect(url_for("register"))

    content = """
    <section class="form-wrap">
      <div class="eyebrow">JOIN THE DIRECTORY</div>
      <h1>Create your profile</h1>
      <p>New agent profiles remain private until reviewed.</p>
      <form class="form-card" method="post">
        <label>Full name
          <input name="name" maxlength="120" required>
        </label>
        <label>City
          <input name="city" maxlength="100" required>
        </label>
        <label>Languages spoken
          <input name="languages" placeholder="Telugu, English" required>
        </label>
        <label>Years of experience
          <input name="experience" type="number"
                 min="0" max="70" value="0" required>
        </label>
        <label>Phone number
          <input name="phone" maxlength="30" required>
        </label>
        <label>Email address
          <input name="email" type="email" maxlength="150" required>
        </label>
        <label>Services offered
          <input name="services" maxlength="250" required>
        </label>
        <label>Professional introduction
          <textarea name="bio" rows="4" maxlength="1000"
                    required></textarea>
        </label>
        <button type="submit">Submit for review →</button>
      </form>
    </section>
    """
    return page("Join as an agent", content)


@app.route("/health")
def health():
    return {"status": "ok", "database": str(DB_PATH)}


print("Website backend and database are ready!")
print("Project folder:", PROJECT)
print("Database file:", DB_PATH)
```

```python
import threading
import time
import webbrowser
import urllib.request

def run_website():
    app.run(
        host="127.0.0.1",
        port=5000,
        debug=False,
        use_reloader=False
    )

server_thread = threading.Thread(
    target=run_website,
    daemon=True
)
server_thread.start()

time.sleep(2)

response = urllib.request.urlopen(
    "http://127.0.0.1:5000/health"
)
print(response.read().decode())

webbrowser.open("http://127.0.0.1:5000")
```
